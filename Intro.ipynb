```python
# ============================================================
# 1. IMPORTAR LAS LIBRERÍAS
# ============================================================

# NetworkX se utiliza para crear y trabajar con redes o grafos.
import networkx as nx

# Matplotlib se utiliza para visualizar y dibujar la red.
import matplotlib.pyplot as plt


# ============================================================
# 2. CREAR LA RED
# ============================================================

# Creamos un grafo dirigido (DiGraph).
# "Dirigido" significa que las conexiones tienen una dirección:
# por ejemplo, 1 -> 2 no necesariamente significa que exista 2 -> 1.
G = nx.DiGraph()


# ============================================================
# 3. AGREGAR LOS NODOS
# ============================================================

# Agregamos los nodos de nuestra red.
# En este caso tenemos 7 nodos, numerados del 1 al 7.
G.add_nodes_from([1, 2, 3, 4, 5, 6, 7])


# ============================================================
# 4. AGREGAR LAS CONEXIONES Y SUS DISTANCIAS
# ============================================================

# Cada tupla tiene la forma:
#
# (nodo_origen, nodo_destino, peso)
#
# Por ejemplo:
# (1, 2, 5)
# significa que existe una conexión de 1 hacia 2
# y que su distancia o costo es 5.

G.add_weighted_edges_from([
    (1, 2, 5),(1, 3, 1),(3, 2, 2),(2, 6, 6),(6, 2, 7),(2, 4, 7),(2, 5, 1),(3, 4, 6),(4, 3, 7),
    (3, 5, 7),(4, 6, 4),(4, 7, 6),(5, 4, 3),(5, 6, 5),(5, 7, 9),(6, 7, 2)])

# ============================================================
# 5. DEFINIR LA POSICIÓN DE CADA NODO
# ============================================================

# Aquí indicamos manualmente dónde queremos colocar
# cada nodo en la gráfica.
#
# Cada nodo tiene una posición (x, y).
#
# Esto NO cambia las conexiones de la red.
# Únicamente sirve para que el dibujo quede ordenado
# y sea más fácil de interpretar.

pos = {
    1: (0, 2),
    2: (2, 4),
    3: (2, 0),
    4: (4, 2),
    5: (6, 0),
    6: (6, 4),
    7: (8, 2)
}


# ============================================================
# 6. CREAR EL ÁREA DE LA FIGURA
# ============================================================

# figsize=(14, 8) define el tamaño de la gráfica.
# El primer número representa el ancho y el segundo la altura.
plt.figure(figsize=(14, 8))


# ============================================================
# 7. DIBUJAR LOS NODOS
# ============================================================

nx.draw_networkx_nodes(
    G,
    pos,

    # Color de los nodos.
    node_color="blue",

    # Color del borde de cada nodo.
    edgecolors="black",

    # Tamaño de los nodos.
    node_size=1300,

    # Grosor del borde de los nodos.
    linewidths=2
)


# ============================================================
# 8. MOSTRAR EL NÚMERO DE CADA NODO
# ============================================================

# Esta función coloca las etiquetas (1, 2, 3, etc.)
# dentro de los nodos.

nx.draw_networkx_labels(
    G,
    pos,

    # Tamaño de la letra.
    font_size=16,

    # Hace que los números aparezcan en negritas.
    font_weight="bold"
)


# ============================================================
# 9. DIBUJAR LAS CONEXIONES
# ============================================================

nx.draw_networkx_edges(
    G,
    pos,

    # True indica que queremos mostrar las flechas.
    # Esto es importante porque nuestra red es dirigida.
    arrows=True,

    # Define el estilo de la punta de las flechas.
    arrowstyle="-|>",

    # Tamaño de las puntas de las flechas.
    arrowsize=20,

    # Color de las conexiones.
    edge_color="black",

    # Grosor de las conexiones.
    width=2,

    # Hace que las conexiones sean rectas.
    # rad=0.0 significa que no tendrán curvatura.
    connectionstyle="arc3,rad=0.0",

    # Evita que la flecha salga exactamente desde el borde
    # del nodo y mejora la visualización.
    min_source_margin=15,
    min_target_margin=15
)


# ============================================================
# 10. OBTENER LOS VALORES DE LAS DISTANCIAS
# ============================================================

# Extraemos los pesos que asignamos anteriormente
# a cada conexión.
#
# Por ejemplo:
# 1 -> 2 = 5
# 1 -> 3 = 1
# 2 -> 5 = 1
#
# Estos valores serán utilizados como etiquetas
# sobre las conexiones.

edge_labels = nx.get_edge_attributes(G, "weight")


# ============================================================
# 11. MOSTRAR LAS DISTANCIAS EN LA GRÁFICA
# ============================================================

nx.draw_networkx_edge_labels(
    G,
    pos,

    # Utilizamos las distancias obtenidas anteriormente.
    edge_labels=edge_labels,

    # Tamaño de los números.
    font_size=14,

    # Coloca la etiqueta aproximadamente a la mitad
    # de cada conexión.
    label_pos=0.5,

    # Evita que el texto rote junto con la conexión.
    # Así todas las distancias se leen horizontalmente.
    rotate=False
)


# ============================================================
# 12. QUITAR LOS EJES
# ============================================================

# Como estamos mostrando una red y no una gráfica matemática,
# no necesitamos mostrar los ejes X y Y.
plt.axis("off")


# ============================================================
# 13. MOSTRAR LA RED
# ============================================================

# Finalmente mostramos la red en pantalla.
plt.show()
```

```python
# ============================================================
# 1. IMPORTAR LAS LIBRERÍAS
# ============================================================

# NetworkX nos permite crear y analizar redes o grafos.
import networkx as nx

# Matplotlib nos permite visualizar gráficamente la red.
import matplotlib.pyplot as plt


# ============================================================
# 2. CREAR LA RED
# ============================================================

# Creamos un grafo NO dirigido.
#
# A diferencia de nx.DiGraph(), aquí las conexiones
# no tienen dirección.
#
# Por ejemplo:
# 1 -- 2
#
# significa que podemos ir de 1 a 2 y de 2 a 1
# utilizando la misma conexión.
G = nx.Graph()


# ============================================================
# 3. AGREGAR LAS CONEXIONES Y SUS PESOS
# ============================================================

# Cada elemento tiene la forma:
#
# (nodo 1, nodo 2, peso)
#
# El peso representa la distancia, costo o valor
# asociado a cada conexión.

G.add_weighted_edges_from([(1, 2, 5),(1, 3, 1),(2, 3, 2),(2, 5, 1),(2, 6, 6),(2, 4, 7),(2, 5, 1),
    (3, 4, 6),(3, 5, 7),(4, 6, 4),(4, 7, 6),(4, 5, 3),(5, 6, 5),(5, 7, 9),(6, 7, 2)])

# ============================================================
# 4. DEFINIR LA POSICIÓN DE LOS NODOS
# ============================================================

# Aquí indicamos manualmente dónde queremos colocar
# cada nodo dentro de la gráfica.
#
# Las coordenadas tienen la forma:
# (x, y)
#
# Estas posiciones solamente afectan la apariencia
# de la gráfica; no modifican la red.

pos = {
    1: (0, 2),
    2: (2, 4),
    3: (2, 0),
    4: (4, 2),
    5: (6, 0),
    6: (6, 4),
    7: (8, 2)
}


# ============================================================
# 5. CREAR LA FIGURA
# ============================================================

# Definimos el tamaño de la ventana donde se dibujará
# nuestra red.
plt.figure(figsize=(12, 7))


# ============================================================
# 6. DIBUJAR LA RED ORIGINAL
# ============================================================

nx.draw(
    G,
    pos,

    # Muestra el número de cada nodo.
    with_labels=True,

    # Color de relleno de los nodos.
    node_color="white",

    # Color del borde de los nodos.
    edgecolors="black",

    # Tamaño de los nodos.
    node_size=1200,

    # Tamaño de los números de los nodos.
    font_size=14,

    # Grosor de las conexiones.
    width=2
)


# ============================================================
# 7. OBTENER LOS PESOS DE LAS CONEXIONES
# ============================================================

# Extraemos los pesos que tienen las conexiones
# para poder mostrarlos en la gráfica.
#
# Por ejemplo:
# 1 -- 3 = 1
# 2 -- 5 = 1
# 6 -- 7 = 2

edge_labels = nx.get_edge_attributes(G, "weight")


# ============================================================
# 8. MOSTRAR LOS PESOS SOBRE LAS CONEXIONES
# ============================================================

nx.draw_networkx_edge_labels(
    G,
    pos,

    # Utilizamos los pesos obtenidos anteriormente.
    edge_labels=edge_labels,

    # Tamaño de los números.
    font_size=12
)


# ============================================================
# 9. AGREGAR TÍTULO Y QUITAR LOS EJES
# ============================================================

plt.title("Red no dirigida")

# No necesitamos mostrar los ejes X y Y,
# ya que estamos trabajando con una red.
plt.axis("off")


# Mostrar la gráfica.
plt.show()


# ============================================================
# 10. CALCULAR EL ÁRBOL DE EXPANSIÓN MÍNIMA
# ============================================================

# minimum_spanning_tree() busca un conjunto de conexiones
# que permita conectar TODOS los nodos de la red,
# utilizando el menor peso total posible.
#
# En otras palabras:
#
# - Todos los nodos deben quedar conectados.
# - No debe haber ciclos.
# - La suma de los pesos debe ser mínima.
#
# "weight='weight'" indica que para realizar el cálculo
# debemos utilizar los pesos de las conexiones.

T = nx.minimum_spanning_tree(
    G,
    weight="weight"
)


# ============================================================
# 11. MOSTRAR LAS ARISTAS DEL ÁRBOL
# ============================================================

print("ARISTAS DEL ÁRBOL DE EXPANSIÓN MÍNIMA")
print("--------------------------------------")


# Recorremos todas las conexiones que forman parte
# del árbol de expansión mínima.
#
# u = primer nodo
# v = segundo nodo
# data = información de la conexión, incluyendo su peso.

for u, v, data in T.edges(data=True):

    # Mostramos los dos nodos y el peso de la conexión.
    print("Nodo", u, "-", v, ": peso =", data["weight"])


# ============================================================
# 12. CALCULAR EL PESO TOTAL
# ============================================================

# Sumamos los pesos de todas las conexiones
# que forman parte del árbol de expansión mínima.

peso_total = T.size(weight="weight")


# Mostramos el resultado.
print("--------------------------------------")
print("Peso total del árbol:", peso_total)


# ============================================================
# 13. CREAR UNA NUEVA FIGURA
# ============================================================

# Ahora vamos a dibujar solamente el Árbol
# de Expansión Mínima.
plt.figure(figsize=(12, 7))


# ============================================================
# 14. DIBUJAR EL ÁRBOL DE EXPANSIÓN MÍNIMA
# ============================================================

nx.draw(
    T,
    pos,

    # Mostrar los números de los nodos.
    with_labels=True,

    # Color de los nodos.
    node_color="white",

    # Color del borde.
    edgecolors="black",

    # Tamaño de los nodos.
    node_size=1200,

    # Tamaño de los números.
    font_size=14,

    # Hacemos las conexiones un poco más gruesas
    # para distinguirlas de la red original.
    width=3
)


# ============================================================
# 15. OBTENER LOS PESOS DEL ÁRBOL
# ============================================================

# Obtenemos únicamente los pesos de las conexiones
# que fueron seleccionadas para formar el árbol.

tree_labels = nx.get_edge_attributes(
    T,
    "weight"
)


# ============================================================
# 16. MOSTRAR LOS PESOS DEL ÁRBOL
# ============================================================

nx.draw_networkx_edge_labels(
    T,
    pos,

    # Utilizamos los pesos de las conexiones del árbol.
    edge_labels=tree_labels,

    # Tamaño de los números.
    font_size=12
)


# ============================================================
# 17. MOSTRAR EL RESULTADO
# ============================================================

plt.title("Árbol de Expansión Mínima")

# Quitamos los ejes para que se vea solamente la red.
plt.axis("off")

# Mostrar la gráfica.
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt
G = nx.DiGraph()
G.add_weighted_edges_from([
    (1, 2, 5),(1, 3, 1),(2, 6, 6),(2, 5, 1),(2, 4, 7),(3, 2, 2),(3, 4, 6),(3, 5, 7),(4, 6, 4)(4, 3, 7),
    (4, 7, 6),(5, 6, 5),(5, 4, 3),(5, 7, 9),(6, 2, 7),(6, 7, 2)])
pos = {
    1: (0, 2),
    2: (2, 4),
    3: (2, 0),
    4: (4, 2),
    5: (6, 0),
    6: (6, 4),
    7: (8, 2)
}

plt.figure(figsize=(12, 7))

nx.draw(
    G,
    pos,
    with_labels=True,
    node_color="white",
    edgecolors="black",
    node_size=1200,
    font_size=14,
    font_weight="bold",
    arrows=True,
    arrowstyle="-|>",
    arrowsize=20,
    width=2
)

edge_labels = nx.get_edge_attributes(G, "weight")

nx.draw_networkx_edge_labels(
    G,
    pos,
    edge_labels=edge_labels,
    font_size=12,
    rotate=False
)

plt.title("Red dirigida")
plt.axis("off")
plt.show()
origen = 1
destino = 7
ruta = nx.shortest_path(
    G,
    source=origen,
    target=destino,
    weight="weight"
)

distancia = nx.shortest_path_length(
    G,
    source=origen,
    target=destino,
    weight="weight"
)
print("RUTA MÁS CORTA")
print("--------------------------------------")
print("Nodo de origen:", origen)
print("Nodo de destino:", destino)
print("Ruta más corta:", ruta)
print("Distancia total:", distancia)


# ==========================================
# 9. OBTENER LAS CONEXIONES DE LA RUTA
# ==========================================

print("--------------------------------------")
print("Conexiones de la ruta:")

for i in range(len(ruta) - 1):
    u = ruta[i]
    v = ruta[i + 1]
    peso = G[u][v]["weight"]

    print(f"{u} -> {v} : peso = {peso}")


# ==========================================
# 10. DIBUJAR LA RUTA MÁS CORTA
# ==========================================

plt.figure(figsize=(12, 7))

# Dibujar todos los nodos
nx.draw_networkx_nodes(
    G,
    pos,
    node_color="white",
    edgecolors="black",
    node_size=1200,
    linewidths=2
)

# Dibujar etiquetas
nx.draw_networkx_labels(
    G,
    pos,
    font_size=14,
    font_weight="bold"
)

# Dibujar todas las conexiones en gris
nx.draw_networkx_edges(
    G,
    pos,
    edge_color="gray",
    arrows=True,
    arrowstyle="-|>",
    arrowsize=18,
    width=1.5,
    alpha=0.5
)

# Crear las conexiones que pertenecen a la ruta
aristas_ruta = []

for i in range(len(ruta) - 1):
    aristas_ruta.append(
        (ruta[i], ruta[i + 1])
    )


# Dibujar la ruta más corta
nx.draw_networkx_edges(
    G,
    pos,
    edgelist=aristas_ruta,
    edge_color="red",
    arrows=True,
    arrowstyle="-|>",
    arrowsize=22,
    width=4
)


# Mostrar pesos de todas las conexiones
edge_labels = nx.get_edge_attributes(
    G,
    "weight"
)

nx.draw_networkx_edge_labels(
    G,
    pos,
    edge_labels=edge_labels,
    font_size=11,
    rotate=False
)

plt.title("Ruta más corta del nodo 1 al nodo 7")
plt.axis("off")
plt.show()
```
